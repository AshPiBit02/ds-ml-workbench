```python
import pandas as pd

orders = pd.DataFrame({
    "order_id": [1001, 1002, 1003, 1004],
    "quantity": [2, 1, 4, 3],
    "unit_price": [25.0, 80.0, 15.0, 40.0],
})
```

```python
# Inspect the shape, columns, and dtypes.
print(f"Shape: {orders.shape}\nColumns: {orders.columns}\ndtypes: \n{orders.dtypes}")
```

```python
orders["revenue"]=orders["quantity"]*orders["unit_price"]
print("Created a 'revenue' column: \n",orders)
```

```python
total_revenue=orders["revenue"].sum()
print("Total revenue: $",total_revenue)
```

```python
# Orders with revenue greater than 50.
filtered_orders=orders[orders["revenue"]>50]
print("Orders with revenue greater than 50: \n",filtered_orders)
```

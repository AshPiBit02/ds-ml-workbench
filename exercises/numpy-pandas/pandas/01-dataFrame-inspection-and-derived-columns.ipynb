```python
# Build a small order table for inspection and calculations
import pandas as pd

orders = pd.DataFrame({
    "order_id": [1001, 1002, 1003, 1004],
    "quantity": [2, 1, 4, 3],
    "unit_price": [25.0, 80.0, 15.0, 40.0],
})
```

```python
# Inspect dimensions, column labels, and data types
# Inspect the shape, columns, and dtypes.
print(f"Shape: {orders.shape}\nColumns: {orders.columns}\ndtypes: \n{orders.dtypes}")
```

```python
# Derived columns are calculated element by element
orders["revenue"]=orders["quantity"]*orders["unit_price"]
print("Created a 'revenue' column: \n",orders)
```

```python
# Sum the revenue column to get total revenue
total_revenue=orders["revenue"].sum()
print("Total revenue: $",total_revenue)
```

```python
# Use a boolean mask to keep orders above the revenue threshold
# Orders with revenue greater than 50.
filtered_orders=orders[orders["revenue"]>50]
print("Orders with revenue greater than 50: \n",filtered_orders)
```

```python
# Add a summary row using the DataFrame index label
orders.loc["Total"]=[
    None,
    orders["quantity"].sum(),
    None,
    orders["revenue"].sum()
]
"""
orders.loc["Total"]={
        "order_id":None,
        "quantity":orders["quantity"].sum(),
        "unit_price":None,
        "revenue":orders["revenue"].sum()
}
"""
print("Final orders: \n",orders)
```

### Important notes

- `shape` is `(rows, columns)`, while `dtypes` shows the type of each column.
- Assigning a new column changes the DataFrame in place; use `.copy()` when an independent DataFrame is needed.
- Boolean filtering returns selected rows; use `.loc[...]` when assigning into a filtered DataFrame to avoid chained-assignment issues.
- Adding a summary row with `None` can promote numeric columns to floating-point because missing values are represented as `NaN`.
- Keep identifiers, quantities, prices, and totals in appropriate columns and validate them before aggregation.

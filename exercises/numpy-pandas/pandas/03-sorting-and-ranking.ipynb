```python
import pandas as pd
products = pd.DataFrame({
    "product": ["A", "B", "C", "D", "E"],
    "units_sold": [120, 80, 150, 95, 150],
    "return_rate": [0.04, 0.02, 0.08, 0.03, 0.05],
})
```

```python
products.sort_values("units_sold",ascending=False)
```

```python
products.sort_values(by=["units_sold","return_rate"],ascending=[False,True])
```

```python
products["rank"]=products["units_sold"].rank(
    ascending=False,
    method="dense").astype(int) # ties are handled using method which is average by default
print("Products with rank: \n",products)
```

```python
products.nsmallest(2,"rank")
```

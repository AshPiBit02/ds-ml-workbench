```python
import pandas as pd
customers = pd.DataFrame({
    "customer_id": ["C01", "C02", "C03", "C04"],
    "name": ["Asha", "Ravi", "Mina", "Noah"],
    "region": ["North", "South", "North", "West"],
    "orders": [5, 2, 8, 3],
})
customers=customers.set_index("customer_id")
```

```python
print("Name & region of customers: \n",customers[["name","region"]])
```

```python
print("Customers from the North region: \n",customers[customers["region"]=="North"])
```

```python
print("Selected the first two rows: \n",customers.iloc[:2])
```

```python
print("Selected the row for customer C03: \n",customers.loc["C03"])
```

| Feature | `.loc` | `.iloc` |
|----------|---------|---------|
| Selects by | Labels (names) | Integer positions |
| Row selection | Index labels | Row numbers (0, 1, 2, ...) |
| Column selection | Column names | Column positions (0, 1, 2, ...) |
| Example | `df.loc[0, "name"]` | `df.iloc[0, 1]` |
| Multiple columns | `df.loc[:, ["name", "region"]]` | `df.iloc[:, [1, 2]]` |
| Slice end | Included | Excluded |
| Best used when | You know labels/names | You know positions |

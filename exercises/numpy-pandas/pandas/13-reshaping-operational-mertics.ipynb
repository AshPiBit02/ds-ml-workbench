```python
# Create monthly operational metrics for each support team
import pandas as pd
metrics = pd.DataFrame({
    "team": ["A", "A", "B", "B"],
    "month": ["Jan", "Feb", "Jan", "Feb"],
    "tickets_closed": [80, 95, 70, 88],
    "average_resolution_hours": [12, 10, 15, 13],
})
metrics
```

```python
# Reshape tickets from long format to one month per column
tickets_closed_wide=metrics.pivot(
    index="team",
    columns="month",
    values="tickets_closed"
)
tickets_closed_wide
```

```python
# Reshape resolution time using the same team-month layout
avg_resolution_hours_wide=metrics.pivot(
    index="team",
    columns="month",
    values="average_resolution_hours"
)
avg_resolution_hours_wide
```

```python
# Convert the wide table back to tidy long format
tickets_closed_long=pd.melt(
    tickets_closed_wide.reset_index(),
    id_vars="team",
    var_name="month",
    value_name="tickets_closed"
)
tickets_closed_long
```

```python
# Verify that reshaping preserves the total number of tickets
wide_tickets_closed_total=tickets_closed_wide["Jan"].sum()+tickets_closed_wide["Feb"].sum()
long_tickets_closed_total=tickets_closed_long["tickets_closed"].sum()
print(f"Tickets closed total: \nBefore reshaping: {wide_tickets_closed_total}\nAfter reshaping: {long_tickets_closed_total}")
```

### When Long Format Is More Useful Than Wide Format

| Long Format Is Better When... | Reason |
|-------------------------------|--------|
| Performing aggregations with `groupby()` | Grouping columns are easier to work with. |
| Creating visualizations | Most plotting libraries expect data in long format. |
| Filtering and sorting by categories | Categories are stored in a single column instead of multiple columns. |
| Working with time-series data | Dates or periods can be stored in one column and analyzed consistently. |
| Applying statistical calculations | Many analytical functions operate more naturally on long-form data. |
| Using `pivot()`, `pivot_table()`, and `groupby()` workflows | These operations typically start from long-format data. |
| Storing data with a variable number of categories | New categories become rows instead of requiring new columns. |
| Following tidy data principles | Each variable has its own column and each observation has its own row. |

### Conclusion

Long format is generally preferred for data analysis, filtering, aggregation, visualization, and machine learning workflows, while wide format is often easier for reporting, dashboards, and human-readable summaries.

### Important notes

- `pivot()` requires each `index` and `columns` combination to be unique; use `pivot_table()` with an aggregation function when duplicates exist.
- Wide format is convenient for side-by-side reporting, while long format is usually easier for grouping, filtering, plotting, and time-based analysis.
- `melt()` turns selected value columns into rows and keeps identifier columns such as `team` unchanged.
- Check totals and row counts before and after reshaping to catch dropped, duplicated, or missing observations.
- Missing combinations may appear as `NaN` after a pivot; decide whether they represent missing data or a true zero before filling them.

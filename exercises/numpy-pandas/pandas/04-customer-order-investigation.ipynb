```python
# Create order data with customer, status, quantity, and price
import pandas as pd
orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105, 106, 107],
    "customer_id": ["C01", "C02", "C01", "C03", "C02", "C04", "C01"],
    "status": ["shipped", "cancelled", "shipped", "pending",
               "shipped", "shipped", "shipped"],
    "quantity": [2, 1, 3, 2, 4, 1, 5],
    "unit_price": [20, 50, 15, 40, 25, 100, 10],
})
orders=orders.set_index("order_id")
```

```python
# Inspect the raw order records
orders
```

```python
# Calculate revenue for each order
orders["revenue"]=orders["quantity"]*orders["unit_price"]
orders
```

```python
# Keep only completed orders for actual revenue analysis
orders=orders[orders["status"].isin(["shipped", "delivered"])]
orders
```

```python
# Aggregate completed revenue by customer
revenue_per_customer=orders.groupby("customer_id")["revenue"].sum()
print("Total revenue per customer: \n",revenue_per_customer)
```

```python
# Find the customer with the largest total revenue
print("Largest valid revenue customer: ",revenue_per_customer.idxmax())
```

```python
# Count orders by status
orders["status"].value_counts()
```

Cancelled orders should not be included in completed revenue because the sale was not successfully completed. When an order is cancelled, the customer does not receive the product and the business does not earn the payment from that transaction.

Including cancelled orders in revenue calculations would:

- Overstate the company's actual earnings.
- Give an inaccurate picture of sales performance.
- Lead to incorrect business decisions based on inflated revenue figures.
- Make financial reports less reliable.

Therefore, only completed orders (such as shipped or delivered orders) should be counted when calculating actual revenue.

### Important notes

- Revenue should include only completed statuses such as `shipped` or `delivered`; pending and cancelled orders should be excluded.
- `groupby()` returns one result per group and excludes missing group keys by default.
- `idxmax()` returns the label of the first row with the maximum value when there is a tie.
- `value_counts()` counts non-missing values by default; use `dropna=False` when missing statuses must be reported.
- Validate quantities, prices, and status values before using them in financial summaries.

```python
# Create sample data containing a potential outlier
import numpy as np
data = np.array([12, 18, 19, 21, 25, 25, 101, 28, 30])
```

```python
# The mean can be pulled upward by extreme values
print("Mean of data: ",data.mean())
```

```python
# The median is more robust to extreme values
print("Median of data: ",np.median(data))
```

```python
# Use the IQR rule to identify values outside the typical range
Q1=np.percentile(data,25)
Q3=np.percentile(data,75)

IQR=Q3-Q1

# Calculating Bounds
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR

# Identify Outliers
outliers=data[(data<lower)|(data>upper)]
print("Outliers: ",outliers)
```

```python
# Keep values inside the IQR-based bounds
clean_data=data[(data>lower)&(data<upper)]
print("Cleaned data (removed outliers): ",clean_data)
```

```python
# A z-score measures how many standard deviations a value is from the mean
mean = np.mean(data)
std = np.std(data)

z = (data - mean) / std

x_clean = data[np.abs(z) < 3]
print(x_clean)
```

```python
# Compare the mean before and after removing detected outliers
print("Original data mean: ",data.mean())
print("Cleaned data mean: ",clean_data.mean())
```

**Important notes**

- Outlier detection should use domain knowledge; an unusual value may be valid and important.
- The IQR rule is often more robust than z-scores because it is less affected by extreme values.
- `np.std()` uses population standard deviation by default; use `ddof=1` for a sample standard deviation.
- This example uses strict bounds, so values exactly equal to a bound are excluded; use `>=` or `<=` when they should be retained.
- Record the cleaning rule and number of removed rows so the transformation remains reproducible.

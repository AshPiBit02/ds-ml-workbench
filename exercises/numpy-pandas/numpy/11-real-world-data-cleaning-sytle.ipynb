```python
import numpy as np
data = np.array([12, 18, 19, 21, 25, 25, 101, 28, 30])
```

```python
print("Mean of data: ",data.mean())
```

```python
print("Median of data: ",np.median(data))
```

```python
# IQR Method to Identify Outliers.
Q1=np.percentile(data,25)
Q3=np.percentile(data,75)

IQR=Q3-Q1

# Calculating Bounds
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR

# Identify Outliers
outliers=data[(data<lower)|(data>upper)]
print("Outliers: ",outliers)
```

```python
# Removing outliers
clean_data=data[(data>lower)&(data<upper)]
print("Cleaned data (removed outliers): ",clean_data)
```

```python
# Removing outlier using Z-Score method
mean = np.mean(data)
std = np.std(data)

z = (data - mean) / std

x_clean = data[np.abs(z) < 3]
print(x_clean)
```

```python
print("Original data mean: ",data.mean())
print("Cleaned data mean: ",clean_data.mean())
```

```python
# Create 1D and 2D arrays for the indexing examples
import numpy as np
arr=np.array([5,1,9,3,7,2])
matrix=np.array([[8,1,6],[3,5,7],[4,9,2]])
```

```python
# Integer-array indexing selects several elements at once (fancy indexing)
indices=np.array([0,2,4])
print("Selected element via indices: ",arr[indices])
```

```python
# np.sort() returns a sorted copy; arr.sort() changes arr in place
print("Sorted arr in ascending: ",np.sort(arr))
```

```python
# Reverse the ascending result to get descending order
print("Sorted arr in descending: ",np.sort(arr)[::-1])
```

```python
# np.argmax() returns the position of the largest value
max_index=np.argmax(arr)
print(f"Index of the maximum value({arr[max_index]}) in arr: ",max_index)
```

```python
# np.argmin() returns the position of the smallest value
min_index=np.argmin(arr)
print(f"Index of the minimum value({arr[min_index]}) in arr: ",min_index)
```

```python
# np.argsort() returns indices that would sort the array
sorted_index=np.argsort(arr) # returns the indices that sort arr
print("Sorted arr: ",arr[sorted_index])
```

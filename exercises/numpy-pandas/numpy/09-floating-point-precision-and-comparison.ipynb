```python
# Create arrays containing decimal values that are not always exact in binary
import numpy as np
x = np.array([0.1, 0.2, 0.3])
y = np.array([0.3, 0.3, 0.3])
z = np.array([0.1 + 0.2, 0.3, 0.3])
```

```python
# Inspect the computed sum before comparing it
print("Sum of x: ",x.sum())
```

```python
# Exact equality can fail because of tiny floating-point errors
print("x.sum() == 0.6? ",x.sum()==0.6)
```

```python
# isclose() compares values within a numerical tolerance
print("np.isclose(x.sum(),0.6)? ",np.isclose(x.sum(),0.6))
```

```python
# Display formatting does not change the stored floating-point value
print("z: ",z) # here NumPy tries to make arrays readable so it rounds result of 0.1+0.2 to 0.3
```

```python
# Python exposes the small representation error in this calculation
print(0.1+0.2)
```

### Why does `0.1 + 0.2` differ from `0.3`?

Decimal fractions such as `0.1` and `0.2` usually have no exact finite representation in binary floating point. NumPy stores the nearest representable value, so small errors can appear during calculations.

Use `np.isclose()` for scalar comparisons and `np.allclose()` for array comparisons instead of relying on exact equality.


```python
# Increase display precision to reveal the stored approximation
np.set_printoptions(precision=20)
print(z)
```

```python
# Rounding is useful for presentation, not for replacing comparisons
print("Rounded value: ",np.round(z,3))
```

**Important notes**

- `np.isclose()` uses both relative (`rtol`) and absolute (`atol`) tolerances; choose them based on the scale of the data.
- `np.round()` changes the returned values for display or reporting; it does not fix the underlying representation error.
- Avoid repeatedly rounding intermediate calculations because it can accumulate error. Round at the final presentation step.
- For money, use integer units such as cents or a decimal type when exact decimal arithmetic is required.

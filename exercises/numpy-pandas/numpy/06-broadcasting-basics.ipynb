```python
import numpy as np
row = np.array([[1, 2, 3]])
col = np.array([[10], [20], [30]])
mat = np.array([[1, 2, 3], [4, 5, 6]])
```

```python
print("Computed row +5: ",row+5)
```

```python
print("Computed row + col: ",row+col)
```

```python
print("Adding a 1D array to a 2D array: \n",np.array([10,20,30]+mat))
```

```python
try:
    print("Invalid addition: \n",np.array([1,2,3])+np.array([4,5]))
except ValueError:
    print("Shapes are not broadcastable. incompatible shapes!")
```

```python
import numpy as np
prices = np.array([249, 99, 175, 420, 130, 85])
products = np.array(["A", "B", "C", "D", "E", "F"])
```

```python
sorted_price_asc=np.sort(prices)
sorted_price_desc=np.sort(prices)[::-1]
print("Sorted prices array in asc: ",sorted_price_asc)
print("Sorted prices array in desc: ",sorted_price_desc)
```

```python
sorted_price_idx_asc=np.argsort(prices)
sorted_price_idx_desc=np.argsort(prices)[::-1]
print("Sorted(asc) prices array index: ",sorted_price_idx_asc)
print("Sorted(desc) prices array index: ",sorted_price_idx_desc)
```

```python
print("Sorted prices array using sorted index in asc: ",prices[sorted_price_idx_asc])
print("Sorted prices array using sorted index in desc: ",prices[sorted_price_idx_desc])
```

```python
print("Cheapest prodcut: ",products[sorted_price_idx_asc[0]])
```

```python
print("Three most expensive products: ",products[sorted_price_idx_desc[:3]])
```

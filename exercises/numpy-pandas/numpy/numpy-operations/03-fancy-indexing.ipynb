```python
import numpy as np
values = np.array([15, 23, 41, 8, 67, 32, 54, 19])
positions = np.array([0, 2, 4, 7])
```

```python
selected_values=values[positions]
print("Selected values at the given positions are: ",selected_values)
```

```python
value_filter=np.array([0,3,-1])
print("Selected first, fourth, and last values using fancy indexing are: ",values[value_filter])
```

```python
print("Reverse the order of the selected positions: ",positions[::-1])
```

```python
repeated_index_array=np.array([0,3,3,5,4,4])
print("Used index array to select repeated positions: ",values[repeated_index_array])
```

Slicing and fancy indexing are both used to select elements from a NumPy array, but they work differently.
* Slicing (arr[start:end]) selects a continuous range of elements.
* Fancy indexing (arr[[1, 3, 5]]) selects specific positions in the array.
* Slicing cannot repeat positions, whereas fancy indexing can repeat indices.
* Slicing preserves the original order of the range, while fancy indexing allows any order of indices.
* Slicing usually returns a view of the original array.
* Fancy indexing returns a copy of the selected data so, the selected result doesn't share any memory with the original array.
* Example:
        values[1:5] → [23, 41, 8, 67] (continuous block)
        values[[1, 4, 2, 4]] → [23, 67, 41, 67] (specific positions with repetition)

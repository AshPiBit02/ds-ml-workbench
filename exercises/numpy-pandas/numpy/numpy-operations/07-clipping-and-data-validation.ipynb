```python
# Create percentage data containing invalid values outside 0-100
import numpy as np
percentages = np.array([-12, 0, 35, 68, 100, 107, 145])
```

```python
# Limit every value to the inclusive range [0, 100]
print("Clipped data: ",np.clip(percentages,0,100))
```

```python
# Identify invalid values before deciding how to handle them
invalid_values=percentages[(percentages>100)|(percentages<0)]
print("Values outside the valid range(0,100): ",invalid_values)
print("Invalid values count: ",len(invalid_values))
```

# Difference Between Clipping and Filtering
   * Filtering removes values that do not satisfy a condition.
   * Clipping keeps all values but replaces out-of-range values with the nearest boundary value.

### Important notes

- `np.clip()` returns a new array and does not modify the original unless the result is assigned back.
- The lower and upper limits are inclusive: values below the lower limit become the lower limit, and values above the upper limit become the upper limit.
- Clipping preserves the number of records but can hide data-quality problems; validate and report invalid values first.
- Filtering removes invalid values, while clipping keeps them and changes their values. Choose based on the meaning of the data.
- Handle missing values separately because comparisons with `NaN` do not identify them as ordinary out-of-range values.

```python
import numpy as np
percentages = np.array([-12, 0, 35, 68, 100, 107, 145])
```

```python
print("Clipped data: ",np.clip(percentages,0,100))
```

```python
invalid_values=percentages[(percentages>100)|(percentages<0)]
print("Values outside the valid range(0,100): ",invalid_values)
print("Invalid values count: ",len(invalid_values))
```

# Difference Between Clipping and Filtering
   * Filtering removes values that do not satisfy a condition.
   * Clipping keeps all values but replaces out-of-range values with the nearest boundary value.

```python
# Create sample temperature data for conditional operations
import numpy as np
temperatures = np.array([12, 19, 24, 31, 8, 27, 35])
```

```python
# Replace values below the lower limit while preserving other values
print("Replaced temperatures below 20 with 20: ",np.where(temperatures<20,20,temperatures))
```

```python
# Convert the condition into categorical labels
labels=np.where(temperatures<20,"cold","warm")
print(labels)
```

```python
# Cap values above the upper limit
print("Replaced temperature above 30 with 30: \n",np.where(temperatures>30,30,temperatures))
```

```python
# Encode the condition as a binary indicator
binary_array=np.where(temperatures>=25,1,0)
print("Binary array: ",binary_array)
```

### Important notes

- `np.where(condition, value_if_true, value_if_false)` returns a new array; it does not modify the original array.
- The true and false values should be broadcastable to the condition shape.
- Conditions such as `<`, `>=`, and `==` produce boolean arrays that can also be used as masks.
- Use `np.select()` when more than two conditions need different outputs.
- Check boundary values carefully: `< 20` excludes `20`, while `<= 20` includes it.

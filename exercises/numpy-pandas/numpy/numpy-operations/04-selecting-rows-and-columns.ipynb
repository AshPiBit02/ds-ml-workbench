````python
import numpy as np
data = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 25, 35],
    [45, 55, 65, 75],
])
````

````python
print("Selected rows 0 and 2: \n",data[[0,2]])
````

````python
print("Selected columns 1 and 3:\n",data[:,[1,3]])
````

````python
# np.ix_() is a NumPy function that creates an open mesh of row and column indices so you can select the Cartesian product (all combinations) of those indices
print("Selected rows 1 and 3 and columns 0 and 2: \n",data[np.ix_([1,3],[0,2])])
````

````python
print("Selected the values at coordinates (0,2), (1,3), and (3,1):\n",data[[0,1,3],[2,3,1]])
````

### `data[[0, 2]]`

- Only **one index array** is provided.
- NumPy treats the array as **row indices**.
- It selects the **entire rows** at positions `0` and `2`.
- Equivalent to:

```python
[data[0], data[2]]
```

---

### `data[[0, 2], [1, 3]]`

- **Two index arrays** are provided.
- NumPy matches the indices **element by element**.
- It forms the coordinate pairs:
  - `(0, 1)`
  - `(2, 3)`
- It returns the elements at those coordinates.
- Equivalent to:

```python
[data[0, 1], data[2, 3]]
```

---

### Core Difference

- `data[[0, 2]]` → Selects **rows** `0` and `2`.
- `data[[0, 2], [1, 3]]` → Selects elements at coordinate pairs **(0,1)** and **(2,3)**.
- With **one index array**, NumPy indexes a single axis (rows).
- With **multiple index arrays**, NumPy combines corresponding indices to form coordinates.

```python
import numpy as np
sales = np.array([
    [120, 150, 130],
    [90, 110, 125],
    [200, 180, 210],
    [160, 155, 170],
])
# Rows represents four weeks. Columns represent three products.
```

```python
total_sales_per_product=sales.sum(axis=0)
print("Total sales for each product: ",total_sales_per_product)
```

```python
total_sales_per_week=sales.sum(axis=1)
print("Total sales for each week: ",total_sales_per_week)
```

```python
avg_sales_per_product=sales.mean(axis=0)
print("Average sales for each product: ",avg_sales_per_product)
```

```python
avg_sales_per_week=sales.mean(axis=1)
print("Average sales for each week: ",np.round(avg_sales_per_week,2))
```

```python
sales=np.sort(sales,axis=1)
print("Sorted each row: \n",sales)
```

```python
sales=np.sort(sales,axis=0)
print("Sorted each column: \n",sales)
```

```python
total_sales_per_week=sales.sum(axis=1)
highest_sales_week=np.argmax(total_sales_per_week)
print(f"Week with the highest sales({total_sales_per_week[highest_sales_week]}): ",highest_sales_week+1)
```

```python
avg_sales_per_product=sales.mean(axis=0)
lowest_sold_product=np.argmin(avg_sales_per_product)
print(f"Product with the lowest average sales({avg_sales_per_product[lowest_sold_product]}): ",lowest_sold_product+1)
```

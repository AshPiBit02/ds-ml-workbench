```python
import numpy as np
scores = np.array([42, 18, 75, 63, 29, 90, 51, 37])
```

```python
passing_mask=scores>=50
print("Mask: ",passing_mask)
```

```python
print("Passing scores: ",scores[passing_mask])
```

```python
print("Passing score count: ",np.sum(passing_mask))
```

```python
mask_30_80=(scores>=30)&(scores<=80)
print("Scores between 30 & 80(inclusive): ",scores[mask_30_80])
```

```python
mask_below_40_or_above_85=(scores<40)|(scores>85)
print("Scores below 40 or above 85: ",scores[mask_below_40_or_above_85])
```

```python
import numpy as np
original = np.array([10, 20, 30, 40, 50, 60])
```

```python
slice=original[1:4]
print("Sliced data: ",slice)
```

```python
slice[0]=999
print("Changed first value of the slice to 99: ",slice)
```

```python
print("Original data: ",original) # original data changed due to the reference from slice array
```

```python
print("slice and original shares memory: ",np.shares_memory(original,slice))
```

```python
original=np.array([10,20,30,40,50,60])
print("Reset data: ",original)
```

```python
fancy_selection=original[[1,2,3]]
print("Selected data using fancy indexing: ",fancy_selection)
```

```python
fancy_selection[0]=99
print("Changed selected data: ",fancy_selection)
print("Original data: ",original) # original data remains unchanged because fancy indexing selection creates copy.
```

```python
copy_data=original.copy()
print("Copied data: ",copy_data)
copy_data[0]=1011
print("Altered copied data: ",copy_data)
print("Original data: ",original)
```

```python
print("copy_data and original shares memory: ",np.shares_memory(original,copy_data))
```

```python
# Create the original array used to demonstrate views and copies
import numpy as np
original = np.array([10, 20, 30, 40, 50, 60])
```

```python
# Basic slicing usually creates a view that shares memory
slice_view=original[1:4]
print("Sliced data: ",slice_view)
```

```python
# Mutating the view can also mutate the original array
slice_view[0]=999
print("Changed first value of the slice_view to 999: ",slice_view)
```

```python
# Check whether changing the slice_view affected the original
print("Original data: ",original) # the original changed because the slice_view is a view
```

```python
# Confirm that both arrays share the same memory
print("slice_view and original shares memory: ",np.shares_memory(original,slice_view))
```

```python
# Reset the original data before the copy examples
original=np.array([10,20,30,40,50,60])
print("Reset data: ",original)
```

```python
# Fancy indexing creates an independent copy of the selected values
fancy_selection=original[[1,2,3]]
print("Selected data using fancy indexing: ",fancy_selection)
```

```python
# Mutating the fancy-indexing result does not affect the original
fancy_selection[0]=99
print("Changed selected data: ",fancy_selection)
print("Original data: ",original) # the original remains unchanged because fancy indexing creates a copy.
```

```python
# .copy() explicitly creates an independent array
copy_data=original.copy()
print("Copied data: ",copy_data)
copy_data[0]=1011
print("Altered copied data: ",copy_data)
print("Original data: ",original)
```

```python
# Confirm that the explicit copy has separate memory
print("copy_data and original shares memory: ",np.shares_memory(original,copy_data))
```

### Important notes

- Basic slicing often returns a view, so edits to the result may change the original array.
- Fancy indexing and boolean indexing return copies, so edits to the result do not change the source array.
- Use `.copy()` when independent data is required and use `np.shares_memory()` to verify memory sharing.
- Not every NumPy operation follows the same view/copy rule; check the operation when mutation matters.
- Avoid naming a variable `slice`, because it shadows Python's built-in `slice` type.

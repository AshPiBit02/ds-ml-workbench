```python
# Create a matrix for locating extreme values
import numpy as np
matrix = np.array([
    [14, 8, 21, 5],
    [7, 19, 3, 12],
    [25, 11, 6, 18],
])
```

```python
# Find the overall maximum and minimum values
print(f"Overall \nMaximum value: {matrix.max()}\nMinimum value: {matrix.min()}")
```

```python
# argmax() returns the flattened position of the maximum
flat_index=np.argmax(matrix)
print("Flattened index of the maximum value: ",flat_index)
```

```python
# Convert a flattened position back to a row-column coordinate
position=np.unravel_index(flat_index,matrix.shape)
print(f"Maximum value position: ({position[0]},{position[1]})")
```

```python
# Find the maximum column position within each row
max_position_row=np.argmax(matrix,axis=1)
print(f"Maximum position in each row: \n1. {max_position_row[0]+1}\n2. {max_position_row[1]+1} \n3. {max_position_row[2]+1}")
```

```python
# Find the minimum row position within each column
min_position_col=np.argmin(matrix,axis=0)
print(f"Minimum position in each column: \n1. {min_position_col[0]+1}\n2. {min_position_col[1]+1} \n3. {min_position_col[2]+1} \n4. {min_position_col[3]+1}")
```

```python
# Build a boolean mask and locate values greater than 15
gt15=matrix>15
print("Values greater than 15: ",matrix[np.nonzero(gt15)])
```

### Important notes

- `np.argmax()` and `np.argmin()` return zero-based indices; add `1` only when displaying one-based positions.
- Without `axis`, a 2D array is treated as flattened. Use `axis=0` for each column and `axis=1` for each row.
- If several values tie for the extreme, these functions return the first matching index.
- `np.unravel_index()` converts a flattened index into coordinates using the array shape.
- `np.nonzero()` returns coordinate arrays; boolean masking is often simpler when only the matching values are needed.


### Related functions

- `np.where(condition)` returns the indices where a condition is true.
- `np.argmax()` / `np.argmin()` locate positions, while `.max()` / `.min()` return values.

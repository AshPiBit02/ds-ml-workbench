```python
# Create scores to classify into multiple categories
import numpy as np
scores = np.array([35, 48, 59, 67, 74, 82, 96])
```

```python
# Apply conditions in order; the first matching condition supplies the label
"""
below 40       -> "fail"
40 through 59  -> "pass"
60 through 79  -> "good"
80 and above   -> "excellent"
"""
categories=np.select([scores>=80,scores>=60,scores>=40],["excellent","good","pass"],default="fail")
print("Scores: ",scores)
print("Score categories: ",categories)
```

```python
# Pair each score with its assigned category for inspection
for score,category in zip(scores,categories):
    print(score," -> ",category)
```

```python
# Count how many scores fall into each category
unique_categories,counts=np.unique(categories,return_counts=True)
for category,count in zip(unique_categories,counts):
    print(f"{category}({count})")
```

### Important notes

- `np.select()` checks conditions from left to right; order overlapping conditions from most specific to least specific.
- Every condition should align with the input shape, and the choice list must have the same length as the condition list.
- `default` handles values that match none of the conditions.
- Use parentheses when combining NumPy conditions, with `&` for AND and `|` for OR; do not use Python `and` or `or` on arrays.
- `np.unique(..., return_counts=True)` returns sorted unique labels and their frequencies.

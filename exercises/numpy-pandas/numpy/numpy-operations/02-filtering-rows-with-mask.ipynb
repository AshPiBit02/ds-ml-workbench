```python
import numpy as np
records = np.array([
    [101, 24, 72],
    [102, 31, 88],
    [103, 19, 61],
    [104, 27, 45],
    [105, 42, 95],
])
"""
Column 0: record ID
Column 1: age
Column 2: score
"""
```

```python
greater_score_mask=records[:,2]>=70
print("Records with scores greater than or equal to 70: \n",records[greater_score_mask])
```

```python
greater_age_mask=records[:,1]>=25
print("Records with age greater than or equal to 25: \n",records[greater_age_mask])
```

```python
combined_mask=greater_age_mask & greater_score_mask
print("Records with scores greater then or equal to 70 and age greater than or equal to 25: \n",records[combined_mask])
```

```python
score_below_50_or_above_90_mask=(records[:,2]<50)|(records[:,2]>90)
print("Records having scores below 50 or above 90 are: \n",records[score_below_50_or_above_90_mask])
```

```python
above_80_mask=records[:,2]>80
print("IDs of records with scores above 80: ",records[above_80_mask][:,0].shape)
```

A filtered two-dimensional array remain two-dimensional.

```python
import numpy as np 
data = np.array([
    [101, 24, 72],
    [102, 31, 88],
    [103, 19, 61],
    [104, 27, 45],
    [105, 42, 95],
    [106, 22, 38],
])
"""
Column 0: record ID
Column 1: age
Column 2: score
"""
```

```python
filter=(data[:,1]>=21) & (data[:,2]>=70)
print("Records with age at least 21 and scores at least 70: \n",data[filter])
```

```python
score=np.where(data[:,2]<50,50,data[:,2])
print("New score array where scores below 50 become 50: \n",score)
```

```python
sorted_record_idx=np.argsort(data[:,2])[::-1]
print("Three highest-scoring record IDs: ",data[sorted_record_idx[:3],0])
```

```python
print("Position of the highest score: ",np.argmax(data[:,2]))
```

```python
clipped_record=data.copy()
clipped_record[:,2]=np.clip(data[:,2],40,80)
print("Clipped record: \n",clipped_record)
```

```python
avg_score=data[:,2].mean()
print("Average score: ",avg_score)
```

```python
above_avg_record=data[data[:,2]>avg_score]
print("Records with score above the average: \n",above_avg_record)
```

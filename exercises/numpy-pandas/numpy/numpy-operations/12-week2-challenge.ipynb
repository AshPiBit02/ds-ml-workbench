```python
# Create records with an ID, age, and score column
import numpy as np 
data = np.array([
    [101, 24, 72],
    [102, 31, 88],
    [103, 19, 61],
    [104, 27, 45],
    [105, 42, 95],
    [106, 22, 38],
])
"""
Column 0: record ID
Column 1: age
Column 2: score
"""
```

```python
# Combine boolean conditions with parentheses and element-wise &
condition_mask=(data[:,1]>=21) & (data[:,2]>=70)
print("Records with age at least 21 and scores at least 70: \n",data[condition_mask])
```

```python
# Apply a minimum score without changing the original data
score=np.where(data[:,2]<50,50,data[:,2])
print("New score array where scores below 50 become 50: \n",score)
```

```python
# Sort score positions in descending order, then select the top three
sorted_record_idx=np.argsort(data[:,2])[::-1]
print("Three highest-scoring record IDs: ",data[sorted_record_idx[:3],0])
```

```python
# Locate the zero-based row position of the highest score
print("Position of the highest score: ",np.argmax(data[:,2]))
```

```python
# Copy the data before clipping scores to the allowed range
clipped_record=data.copy()
clipped_record[:,2]=np.clip(data[:,2],40,80)
print("Clipped record: \n",clipped_record)
```

```python
# Calculate the mean score as a comparison baseline
avg_score=data[:,2].mean()
print("Average score: ",avg_score)
```

```python
# Use a boolean mask to keep scores strictly above the average
above_avg_record=data[data[:,2]>avg_score]
print("Records with score above the average: \n",above_avg_record)
```

### Important notes

- Use parentheses around each NumPy condition; combine array conditions with `&` or `|`, not Python `and` or `or`.
- `np.where()` and `np.clip()` return new values; assigning them to a copy preserves the original records.
- `np.argsort()` returns row positions, so use those positions to retrieve IDs or other aligned columns.
- `np.argmax()` returns the first position when multiple rows share the maximum score.
- Decide whether threshold comparisons should be inclusive (`>=`, `<=`) or strict (`>`, `<`) before filtering.
- Validate that the score column contains appropriate numeric values before calculating averages or applying limits.

```python
import numpy as np
x = np.array([0.1, 0.2, 0.3])
y = np.array([0.3, 0.3, 0.3])
z = np.array([0.1 + 0.2, 0.3, 0.3])
```

```python
print("Sum of x: ",x.sum())
```

```python
print("x.sum() == 0.6? ",x.sum()==0.6)
```

```python
print("np.isclose(x.sum(),0.6)? ",np.isclose(x.sum(),0.6))
```

```python
print("z: ",z) # here NumPy tries to make arrays readable so it rounds result of 0.1+0.2 to 0.3
```

```python
print(0.1+0.2)
```

0.1+0.2 is not exactly 0.3 in binary floating point because computer cannot store binary floating values directly. So it first converts into binary form then due to the non terminating 
nature of the values such as 0.1 & 0.2 python rounds to closest possible value as float can only store upto 64bits. 

```python
np.set_printoptions(precision=20)
print(z)
```

```python
print("Rounded value: ",np.round(z,3))
```

```python
# Create sample sales and score data for summary statistics
import numpy as np
sales = np.array([120, 135, 110, 150, 140, 125, 160, 170])
scores = np.array([5, 7, 9, 8, 10, 6, 8, 7])
```

```python
# Compare center, range, and spread of the sales data
print(f"Sales stats:\n1.Mean: {sales.mean()}\n2.Median: {np.median(sales)}\n3.Min: {sales.min()}\n4.Max: {sales.max()}\n5.Standard Deviation: {sales.std():.2f}")
```

```python
# Compare center, range, and spread of the score data
print(f"Score stats:\n1.Mean: {scores.mean()}\n2.Median: {np.median(scores)}\n3.Min: {scores.min()}\n4.Max: {scores.max()}\n5.Standard Deviation: {scores.std():.2f}")
```

```python
# The 25th percentile: 25% of values are at or below this value
print("Value below which 25 percent of the data falls: ",np.quantile(scores,0.25))
```

```python
# Calculate the first and third quartiles together
print("The 25th & 75th percentiles are: ",np.percentile(scores,[25,75]))
```

A Single extreme value (outlier) affects the mean much more than the median because:
1. Mean uses every value in its calculation.
2. Median depends only on the middle position.

Data:      [10, 20, 30, 40, 50]
Mean = 30, Median = 30

Add outlier: [10, 20, 30, 40, 500]
Mean = 120, Median = 30

**Important notes**

- `np.std()` uses population standard deviation by default (`ddof=0`). Use `ddof=1` for a sample standard deviation.
- The mean is sensitive to outliers; use the median and quartiles for a more robust summary.
- `np.quantile()` and `np.percentile()` use linear interpolation by default when a position falls between values.
- For real-world data, check missing values and units before comparing summary statistics.

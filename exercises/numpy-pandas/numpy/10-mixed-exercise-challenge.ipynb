```python
# Create arrays for arithmetic, broadcasting, aggregation, and filtering
import numpy as np
A = np.array([[1, 2, 3], [4, 5, 6]])
B = np.array([[10, 20, 30], [40, 50, 60]])
C = np.array([2, 4, 6])
D = np.array([0.1, 0.2, 0.3, 0.4])
```

```python
# Matching shapes allow element-wise addition
print("Element wise addition of A & B: \n",A+B)
```

```python
# C broadcasts across each row because its shape matches A's columns
print("Multiply A by C using broadcasting: \n",A*C)
```

```python
# axis=1 reduces each row to one sum
print("Row wise sum of A: ",A.sum(axis=1))
```

```python
# axis=0 reduces each column to one mean
print("Column wise mean of A: ",A.mean(axis=0))
```

```python
# A boolean mask selects values that satisfy the condition
print("Filtered values in D that are greater than 0.25: ",D[D>0.25])
```

```python
# np.where() replaces values conditionally and returns a new array
print("Replaced values in D that are less than 0.3 with 0: ",np.where(D<0.3,0,D))
```

**Important notes**

- Broadcasting works only when trailing dimensions are equal or one of them is `1`.
- `axis=0` operates down rows (one result per column); `axis=1` operates across columns (one result per row).
- Boolean indexing returns a filtered copy, while `np.where()` returns a new array with conditional replacements.
- For floating-point thresholds, use a tolerance such as `np.isclose()` when exact equality is not reliable.

```python
import numpy as np

# Create one-dimensional, two-dimensional, and float arrays.
A=np.array([1,2,3,4,5])
B=np.array([[1,2,3],[4,5,6]])
C=np.array([1,2,3],dtype=np.float64)
```

```python
# .shape returns the size of the array along each dimension.
print("Shape of A: ",A.shape)
print("Shape of B: ",B.shape)
print("Shape of C: ",C.shape)
```

```python
# .ndim reports the number of dimensions in each array.
print("ndim of A: ",A.ndim)
print("ndim of B: ",B.ndim)
print("ndim of C: ",C.ndim)
```

```python
# Reshape A from a one-dimensional array into 5 rows and 1 column.
A=A.reshape(5,1)
print("Reshaped A into a 5x1 array: ",A)
```

```python
# np.arange excludes the stop value, so this creates values from 1 through 6.
print("3x2 array from the numbers 1 to 6 using np.arrange() and reshape: \n",np.arange(1,7).reshape(3,2))
```

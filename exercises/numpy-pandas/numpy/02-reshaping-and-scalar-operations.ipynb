```python
import numpy as np

# Create an array containing values from 0 to 11.
arr=np.arange(12)
```

```python
print("Array: ",arr)
```

```python
# Reshape the 12 values into 3 rows and 4 columns.
arr=arr.reshape(3,4)
print("Reshaped array: \n",arr)
```

```python
# Add 5 to every element using NumPy's element-wise operations.
arr=arr+5
print("arr+5: \n",arr)
```

```python
# Multiply every element by 5.
arr=arr*5
print("arr*5: \n",arr)
```

```python
# Division produces floating-point values in NumPy.
arr=arr/5
print("arr/5: \n",arr)
```

```python
# Reshape the array into 2 blocks, each with 2 rows and 3 columns.
arr=arr.reshape(2,2,3)
print("Reshaped into 2x2x3: \n",arr)
```

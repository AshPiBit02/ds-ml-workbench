```python
import numpy as np
sales = np.array([120, 135, 110, 150, 140, 125, 160, 170])
scores = np.array([5, 7, 9, 8, 10, 6, 8, 7])
```

```python
print(f"Sales stats:\n1.Mean: {sales.mean()}\n2.Median: {np.median(sales)}\n3.Min: {sales.min()}\n4.Max: {sales.max()}\n5.Standard Deviation: {sales.std():.2f}")
```

```python
print(f"Score stats:\n1.Mean: {scores.mean()}\n2.Median: {np.median(scores)}\n3.Min: {scores.min()}\n4.Max: {scores.max()}\n5.Standard Deviation: {scores.std():.2f}")
```

```python
print("Value above which 75 percent of the data falls: ",np.quantile(scores,0.25))
```

```python
print("The 25th & 75th percentiles are: ",np.percentile(scores,[25,75]))
```

A Single extreme value (outlier) affects the mean much more than the median because:
1. Mean uses every value in its calculation.
2. Median depends only on the middle position.

Data:      [10, 20, 30, 40, 50]
Mean = 30, Median = 30

Add outlier: [10, 20, 30, 40, 500]
Mean = 120, Median = 30

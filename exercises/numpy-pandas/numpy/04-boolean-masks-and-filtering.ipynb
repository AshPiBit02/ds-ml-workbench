```python
import numpy as np
scores = np.array([12, 45, 30, 18, 60, 22, 50]) 
temp = np.array([18, 22, 14, 16, 26, 20, 24])
```

```python
# Select scores greater than 30.
mask=scores>30
print("Filtered scores using the mask: ",scores[mask])
```

```python
# Use the temperature mask to select matching scores.
mask2=temp>=20
print("Values in scores that correspond to temp >=20: ",scores[mask2])
```

```python
# Combine conditions with &; wrap each condition in parentheses.
mask3=(scores>=20)&(scores<=50)
print("Scores between 20 & 50: ",scores[mask3])
```

```python
# Replace values below 25 with 0; keep the remaining values unchanged.
new_scores=np.where(scores<25,0,scores)
print("Replaced scores below 25 with 0: ",new_scores)
```

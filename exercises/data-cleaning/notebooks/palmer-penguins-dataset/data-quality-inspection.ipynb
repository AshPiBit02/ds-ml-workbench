```python
# Load the Palmer Penguins dataset and preview its first records
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parents[1]))
from data.raw.penguins import load_penguins
df = load_penguins()
df.head()
```

```python
# Capture the dataset row and column counts
n_rows=df.shape[0]
n_cols=df.shape[1]
print("Rows: ",n_rows)
print("Columns: ",n_cols)
```

```python
# List all available column names
print("Columns name:")
print('-'*30)
df.columns.tolist()
```

```python
# Inspect each column data type
print("Columns dtype:")
print('-'*30)
df.dtypes
```

```python
# Count missing values for every column
missing_values=df.isna().sum().reset_index()
missing_values.columns=["Column Name","Missing Value Count"]
missing_values
```

```python
# Find exact duplicate rows and report how many exist
duplicate_rows=df[df.duplicated(keep="first")]
print("Duplicated Rows: \n",duplicate_rows)
print("Exact duplicated row count: ",len(duplicate_rows))
```

```python
# Classify columns as numerical or categorical
import pandas as pd
column_types=pd.DataFrame({
    "Column Name":df.columns,
    "Type":["Numerical" if pd.api.types.is_numeric_dtype(df[col]) else "Categorical" for col in df.columns]
})
column_types
```

```python
# Count numerical and categorical features
categorical_column_count=(column_types["Type"]=="Categorical").sum()
numerical_column_count=len(column_types)-categorical_column_count
print("Number of Categorical columns: ",categorical_column_count)
print("Number of Numerical columns: ",numerical_column_count)
```

### Important notes

- `isna().sum()` counts missing values in each column; combine it with percentage and row-level counts for a clearer quality report.
- Exact duplicate checks do not find near-duplicates or repeated biological observations; define an entity key and compare it when possible.
- A numerical-looking column may still be categorical, and a categorical column may contain numeric codes; verify meanings from the dataset metadata.
- Use domain knowledge before removing rows or imputing values; missing measurements and missing sex may require different handling.
- Document every cleaning decision, including whether missing values are excluded, filled, or retained for later analysis.
